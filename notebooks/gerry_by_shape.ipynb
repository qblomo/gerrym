```python
import geopandas as gpd 
import pandas as pd 
import matplotlib.pyplot as plt

districts = gpd.read_file("/home/pblomo/Data/Gerrymendering/districts025.shx")
districts['convex_hull'] = districts.convex_hull
this_district = districts.iloc[[0]]
this_ax = this_district['convex_hull'].plot()
this_district['geometry'].plot(ax = this_ax, color = 'red', alpha = 0.5)
```

```python
import matplotlib.pyplot as plt
import shapely

fig, ax = plt.subplots(figsize=(6,6))

geom = this_district.geometry.iloc[0]

gpd.GeoSeries([geom]).plot(ax=ax, color="lightblue", edgecolor="black")
gpd.GeoSeries([geom.minimum_rotated_rectangle]).plot(ax=ax, facecolor="none", edgecolor="red", linewidth=2)
gpd.GeoSeries([shapely.minimum_bounding_circle(geom)]).plot(ax=ax, facecolor="none", edgecolor="green", linewidth=2)

ax.set_aspect("equal")
plt.show()
```
